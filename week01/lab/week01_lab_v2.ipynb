## ส่วน 1 — ตัวช่วย Prompt Log

ทุกครั้งที่ใช้ AI ช่วย ให้เรียก `log_prompt(...)` หนึ่งครั้ง เมื่อจบแล็บจะ export เป็นไฟล์ Markdown แนบท้ายงาน
(ฟังก์ชันนี้อาจารย์เขียนไว้ให้ — ลองอ่านและเขียน Explain This ของฟังก์ชันนี้ใน Markdown ด้านล่างโค้ด)

> ### ✅ ทำตอนนี้ — ส่วน 1 (≈ 8 นาที)
> 1. ▶️ รันเซลล์โค้ดด้านล่าง (ไม่มีข้อความออก = ปกติ ฟังก์ชันถูกสร้างไว้เฉย ๆ)
> 2. 📝 อ่านโค้ดทีละบรรทัด แล้วเขียน *Explain This* ในเซลล์ Markdown ใต้โค้ด
> 3. 👀 อ่าน **ตัวอย่างเต็ม 1 รอบ** ใต้ Explain This — เป็นแบบอย่างของวงจร Prompt → Review → Execute → Verify ที่คุณจะทำซ้ำ 2 ครั้งในวันนี้
>
> **เสร็จเมื่อ:** เขียน Explain This ครบ 3 ข้อ · **ห้ามข้ามเซลล์นี้** เพราะถ้าไม่รัน `log_prompt` ในส่วนถัดไปจะ error

````python
from datetime import datetime

PROMPT_LOG = []   # รายการ Prompt ทั้งหมดของแล็บนี้

def log_prompt(prompt, result, verdict, note=""):
    """บันทึก 1 รอบของการคุยกับ AI
    verdict: "ผ่าน" (ใช้ได้เลย) | "แก้" (ต้องแก้ก่อนใช้) | "ทิ้ง" (ใช้ไม่ได้)
    """
    assert verdict in ("ผ่าน", "แก้", "ทิ้ง"), "verdict ต้องเป็น ผ่าน / แก้ / ทิ้ง"
    PROMPT_LOG.append(dict(time=datetime.now().strftime("%H:%M"),
                           prompt=prompt.strip(), result=result.strip(),
                           verdict=verdict, note=note.strip()))
    print(f"บันทึก Prompt #{len(PROMPT_LOG)} แล้ว ({verdict})")

def _cell(s):
    return s.replace("|", "\\|").replace("\n", "<br>")

def save_prompt_log(path="prompt_log_week01.md"):
    lines = [f"# Prompt Log — สัปดาห์ 1 — {STUDENT_NAME} ({STUDENT_ID})", "",
             f"เครื่องมือ AI: {AI_TOOL}", "",
             "| # | เวลา | Prompt | ผลที่ได้ | Review | หมายเหตุ |", "|---|---|---|---|---|---|"]
    for i, r in enumerate(PROMPT_LOG, 1):
        lines.append(f"| {i} | {r['time']} | {_cell(r['prompt'])} | {_cell(r['result'])} | {r['verdict']} | {_cell(r['note'])} |")
    with open(path, "w", encoding="utf-8") as f:
        f.write("\n".join(lines) + "\n")
    print("บันทึกไฟล์:", path, "| จำนวน prompt:", len(PROMPT_LOG))
````

### Explain This — `log_prompt` / `save_prompt_log`
- **ทำอะไร:** _(เขียนด้วยคำของคุณเอง)_
- **ข้อมูลเปลี่ยนอย่างไร:** _(ที่นี่คือข้อมูลอะไรเข้า/ออก)_
- **จุดเสี่ยง + วิธีตรวจ:** _(เช่น ถ้า verdict พิมพ์ผิดจะเกิดอะไร)_

### 📖 ตัวอย่างเต็ม 1 รอบของวงจร AI (งานสมมติ — **ไม่ใช่ข้อของคุณ** ใช้ดูรูปแบบเท่านั้น)

**โจทย์สมมติ:** ฟังก์ชัน `band_stats(img)` สรุปค่าต่ำสุด/เฉลี่ย/สูงสุดของแต่ละแบนด์สีของ chip

**① Prompt** — เขียนตามเทมเพลตของวิชา (บอก CRS/หน่วย/รูปทรงข้อมูลชัดเจน)

```
[บทบาท]    คุณคือผู้ช่วยเขียน Python สำหรับงานภูมิสารสนเทศ ใช้ Google Colab
[เป้าหมาย] เขียนฟังก์ชัน band_stats(img) สรุปค่าต่ำสุด/เฉลี่ย/สูงสุดของแต่ละแบนด์สี
[ข้อมูลเข้า] img = numpy array ชนิด uint8 รูปทรง (3, H, W) ลำดับแบนด์ R, G, B
[ข้อกำหนดเชิงพื้นที่] ฟังก์ชันนี้ไม่แปลง CRS · 1 พิกเซล = 10 m
[ไลบรารีที่ใช้ได้] numpy เท่านั้น
[รูปแบบผลลัพธ์] dict เช่น {"R": (min, mean, max), "G": ..., "B": ...} ปัดทศนิยม 1 ตำแหน่ง
[ข้อจำกัด] 1 เซลล์ สั้น คอมเมนต์ไทย มีตัวอย่างรันกับ array สุ่ม (3, 4, 4) และพิมพ์ shape
```

**② Review (Explain This) — ก่อนรัน** อ่านโค้ดที่ AI ให้ แล้วเขียนด้วยคำของตัวเอง (ตัวอย่างระดับ "ดี"):

- **ทำอะไร:** วนทีละแบนด์ (แกนแรกของ array) หาค่าต่ำสุด เฉลี่ย สูงสุดของพิกเซลทั้งภาพ ปัดเป็นทศนิยม 1 ตำแหน่ง เก็บเป็น dict ตามชื่อแบนด์
- **ข้อมูลเปลี่ยนอย่างไร:** เข้าเป็น `(3, H, W)` uint8 ออกเป็น dict 3 คู่ ฟังก์ชันไม่แตะ CRS หรือตำแหน่งของภาพ
- **จุดเสี่ยง + วิธีตรวจ:** ถ้าส่งภาพรูปทรง `(H, W, 3)` เข้ามา `img[0]` จะเป็นแถวแรกของภาพ ไม่ใช่แบนด์แดง ผลผิดโดยไม่มี error → ตรวจ `img.shape[0] == 3` ก่อนใช้

**③ Execute** — รันทีละเซลล์ ดู shape และผลที่พิมพ์ออกมา

**④ Verify** — ไม่เชื่อ AI ทดสอบด้วยข้อมูลที่ *เรารู้คำตอบ*:

```python
np.full((3, 2, 2), 100, dtype=np.uint8)   # ทุกพิกเซลเท่ากับ 100 → ทุกแบนด์ควรได้ (100.0, 100.0, 100.0)
```

**⑤ log_prompt** — บันทึกทุกรอบ แม้รอบที่ผิด:

```python
log_prompt(
    prompt="(วาง prompt ทั้งหมดที่ส่งให้ AI)",
    result="รอบแรก AI ใช้ img[..., i] ผิดแกน ได้ shape (3, 4) ผลผิด · รอบสองเปลี่ยนเป็น img[i] ถูกต้อง",
    verdict="แก้",
    note="สั่งให้ใช้ img[i] และเพิ่ม assert img.shape[0] == 3 เอง",
)
```

**verdict เลือกอย่างไร:** `"ผ่าน"` = ใช้ได้เลยโดยไม่แก้ · `"แก้"` = ต้องแก้หรือถามซ้ำก่อนใช้ · `"ทิ้ง"` = ใช้ไม่ได้ ต้องเขียนใหม่

| `fetch_chip(lat, lon, size_m=2240)` | จุดกลาง (องศา) | `(img, item)` — `img` = numpy uint8 รูปทรง `(3, H, W)` (RGB), `item` = STAC item (ต้องมี `item.datetime` และ `item.properties["eo:cloud_cover"]`) |

**เสร็จเมื่อ:** `fetch_chip(16.472, 102.823)` คืนภาพ shape `(3, ~224, ~224)` และ `log_prompt` แล้ว

````python
# 2.1 [โค้ดอ้างอิงจากอาจารย์] ฟังก์ชันค้นหาและอ่านภาพ Sentinel-2 เป็น chip เล็ก ๆ
import numpy as np
import rasterio
from rasterio.windows import from_bounds
from rasterio.warp import transform_bounds
from pystac_client import Client

STAC_URL = "https://earth-search.aws.element84.com/v1"
catalog = Client.open(STAC_URL)

def bbox_from_point(lat, lon, size_m=2240):
    """สร้างกรอบ (west, south, east, north) หน่วยองศา รอบจุดศูนย์กลาง ขนาด size_m เมตร"""
    dlat = (size_m / 2) / 111_320
    dlon = (size_m / 2) / (111_320 * np.cos(np.radians(lat)))
    return tuple(float(x) for x in (lon - dlon, lat - dlat, lon + dlon, lat + dlat))

def fetch_chip(lat, lon, size_m=2240, datetime="2026-01-01/2026-03-31", max_cloud=10):
    """คืน (ภาพ RGB array รูปทรง (3,H,W), STAC item) ของภาพที่เมฆน้อยที่สุดในช่วงเวลาที่กำหนด"""
    bbox = bbox_from_point(lat, lon, size_m)
    search = catalog.search(collections=["sentinel-2-l2a"], bbox=bbox, datetime=datetime,
                            query={"eo:cloud_cover": {"lt": max_cloud}}, max_items=20)
    items = sorted(search.items(), key=lambda it: it.properties["eo:cloud_cover"])
    if not items:
        raise RuntimeError("ไม่พบภาพ — ลองขยายช่วงวันที่หรือเพิ่ม max_cloud")
    item = items[0]
    href = item.assets["visual"].href              # ภาพสีจริง 8-bit ความละเอียด 10 ม.
    with rasterio.open(href) as src:
        # ภาพอยู่ใน UTM แต่กรอบเราเป็นองศา -> ต้องแปลงกรอบเป็น CRS ของภาพก่อนอ่าน
        l, b, r, t = transform_bounds("EPSG:4326", src.crs, *bbox)
        window = from_bounds(l, b, r, t, src.transform)
        img = src.read(window=window)
        print(f"  CRS ภาพ: {src.crs} | shape: {img.shape} | ช่วงค่า: {img.min()}–{img.max()}")
    return img, item
````

### 2.4 ล็อกเฉลย (`truth`) ก่อนเห็นผลโมเดล

กรอก `truth_draft` จากใบงานส่วน D1 (Site Scouting) แล้วรันเซลล์ถัดไปเพื่อ **ล็อก** พร้อมบันทึกเวลา และกรอกคะแนนที่คุณทำนายไว้วันจันทร์ (`PREDICTED_SCORE`, 0–5)

**กติกา:** เซลล์ CLIP (ส่วน 3) จะไม่ทำงานจนกว่าจะล็อกสำเร็จ — ถ้ายังไม่ได้ตรวจ Google Earth ให้ตรวจเดี๋ยวนี้ (5 นาที) ก่อนไปต่อ เหตุผลคือ ถ้าเห็นผลโมเดลก่อน เราจะปรับคำตอบให้เข้ากับโมเดลโดยไม่รู้ตัว

คลาสที่ใช้ได้: `rice paddy field` · `sugarcane field` · `forest` · `water body` · `urban area with buildings` · `bare soil`

> ### ✅ ทำตอนนี้ — 2.4 ล็อก `truth` (≈ 8 นาที)
> 1. ✏️ เปิดใบงานส่วน D1 (ผลที่คุณตรวจ Google Earth เมื่อคืน) แล้วแก้ `None` เป็นชื่อคลาสที่ *คุณเห็น* ครบ 5 จุด
> 2. ✏️ ใส่ `PREDICTED_SCORE` = คะแนนที่คุณทำนายไว้วันจันทร์ (ตัวเลข 0–5)
> 3. ▶️ รันเซลล์ → ควรเห็น `ล็อก truth สำเร็จเมื่อ HH:MM:SS` · **ล็อกแล้วไปดูผลโมเดลได้ ย้อนกลับมาแก้เงียบ ๆ ไม่ได้**
>
> **รูปแบบการกรอก** (ตัวอย่างสมมติ ไม่ใช่เฉลยจริง):
>
> ```python
> truth_draft = {
>     "จุดตัวอย่าง": "forest",        # ต้องเป็นหนึ่งในคลาสที่อนุญาต พิมพ์ให้ตรงตัวอักษร
> }
> PREDICTED_SCORE = 3                  # เป็นตัวเลข ไม่ใส่เครื่องหมายคำพูด
> ```
>
> **ถ้าเจอ error:** `ยังไม่ได้กรอก: [...]` = ยังเหลือ `None` อยู่ · `ชื่อคลาสไม่อยู่ในรายการ` = พิมพ์ชื่อคลาสไม่ตรงกับรายการด้านบน · `กรอก PREDICTED_SCORE เป็นตัวเลข 0–5` = ลืมใส่หรือใส่เป็นข้อความ

````python
# 2.4 ล็อกเฉลยจากการสำรวจ Google Earth (ต้องรันก่อนส่วน 3)
from datetime import datetime

LABELS_ALLOWED = ["rice paddy field", "sugarcane field", "forest", "water body",
                  "urban area with buildings", "bare soil"]       # ต้องตรงกับ LABELS ในส่วน 3

# กรอกจากใบงานส่วน D1 — ใส่คลาสที่ *คุณ* เห็น (ถ้าไม่ใช่อย่างที่ชื่อจุดบอก ให้ใส่ตามที่เห็นจริง)
truth_draft = {
    "มข. (วิทยาเขตหลัก)":        None,
    "บึงแก่นนคร":                None,
    "เขื่อนอุบลรัตน์":            None,
    "ภูเวียง (พื้นที่ป่า)":       None,
    "ที่ราบลุ่มน้ำชี (นาข้าว)":   None,
}
PREDICTED_SCORE = None     # คะแนน CLIP ที่ทำนายไว้ในใบงานส่วน C ข้อ 6 (ตัวเลข 0–5)

def lock_truth(draft, predicted):
    """ตรวจว่ากรอกครบและเป็นคลาสที่อนุญาต แล้วคืนสำเนาที่ล็อกพร้อมเวลา"""
    missing = [k for k, v in draft.items() if v is None]
    bad = {k: v for k, v in draft.items() if v is not None and v not in LABELS_ALLOWED}
    assert not missing, f"ยังไม่ได้กรอก: {missing} — ตรวจ Google Earth ก่อน แล้วค่อยล็อก"
    assert not bad, f"ชื่อคลาสไม่อยู่ในรายการ: {bad}"
    assert predicted is not None and 0 <= predicted <= len(draft), "กรอก PREDICTED_SCORE เป็นตัวเลข 0–5"
    return dict(draft), datetime.now()

truth_locked, TRUTH_LOCKED_AT = lock_truth(truth_draft, PREDICTED_SCORE)
print("ล็อก truth สำเร็จเมื่อ", TRUTH_LOCKED_AT.strftime("%H:%M:%S"), "| คะแนนที่ทำนายไว้:", PREDICTED_SCORE)
````

## ส่วน 3 — Zero-shot Vision Model (CLIP)

**Zero-shot** = ให้โมเดลจำแนกภาพ "โดยไม่ต้องฝึกด้วยตัวอย่างของคลาสนั้นเลย" เราเพียงบอกรายชื่อคลาสเป็นข้อความ
CLIP เรียนจากภาพ+คำบรรยายบนอินเทอร์เน็ตเป็นหลัก ไม่ได้ออกแบบมาสำหรับภาพดาวเทียม 10 เมตร — เราจึงต้องประเมินอย่างระวัง

> ### ✅ ทำตอนนี้ — ส่วน 3 (≈ 12 นาที)
> 1. ▶️ รันเซลล์ **3.1** — ครั้งแรกโหลดโมเดล ~600 MB รอ 1–3 นาที → ควรเห็น `โหลดโมเดลสำเร็จ`
> 2. ▶️ รันเซลล์ **3.2** → ควรเห็น 5 บรรทัดรูปแบบ `ชื่อจุด -> คลาส คะแนน%, คลาส คะแนน%, ...`
> 3. 📝 เขียน Explain This `run_zero_shot`
>
> ⚠️ ถ้าเจอ `ยังไม่ได้ล็อก truth` ให้กลับไปรันเซลล์ **2.4** ก่อน นี่คือกติกาที่ตั้งใจไว้ ไม่ใช่ข้อผิดพลาด
>
> **ข้อควรคิดระหว่างรอ:** คะแนน 90% หมายความว่าโมเดล "มั่นใจ" ไม่ได้แปลว่า "ถูก" — เก็บไว้ตรวจในส่วน 5

````python
# 3.1 โหลดโมเดล zero-shot (ครั้งแรกดาวน์โหลดโมเดล ~600 MB รอสักครู่)
from transformers import pipeline
from PIL import Image

device = 0 if torch.cuda.is_available() else -1
zs = pipeline("zero-shot-image-classification", model="openai/clip-vit-base-patch32", device=device)
print("โหลดโมเดลสำเร็จ")
````

````python
assert "TRUTH_LOCKED_AT" in globals(), "ยังไม่ได้ล็อก truth — กลับไปรันเซลล์ 2.4 ก่อน"
# 3.2 ฟังก์ชันรัน zero-shot กับ chip
def to_pil(img_chw, size=224):
    """array (3,H,W) uint8 -> ภาพ PIL ขนาด size x size (ขยายภาพ 10 ม./พิกเซล ให้พอดีโมเดล)"""
    return Image.fromarray(np.transpose(img_chw, (1, 2, 0))).resize((size, size), Image.BICUBIC)

def run_zero_shot(chips, labels, template="a satellite image of {}."):
    out = {}
    for name, c in chips.items():
        preds = zs(to_pil(c["img"]), candidate_labels=labels, hypothesis_template=template)
        out[name] = preds                       # list ของ dict(label, score) เรียงจากมากไปน้อย
    return out

LABELS = ["rice paddy field", "sugarcane field", "forest", "water body",
          "urban area with buildings", "bare soil"]

results = run_zero_shot(chips, LABELS)
for name, preds in results.items():
    top = ", ".join(f"{p['label']} {p['score']:.0%}" for p in preds[:3])
    print(f"{name:28s} -> {top}")
````

- **ทำอะไร:** _(`hypothesis_template` ทำหน้าที่อะไร?)_
- **ข้อมูลเปลี่ยนอย่างไร:** _(ภาพ 10 ม. ถูกขยายเป็น 224×224 — โมเดลเห็นอะไรที่ต่างจากภาพถ่ายทั่วไป?)_
- **จุดเสี่ยง + วิธีตรวจ:** _(ทำไมคะแนนความมั่นใจสูง ไม่ได้แปลว่าถูก?)_
